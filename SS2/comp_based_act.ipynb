# Bearing and Distance

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Route instructions
bearings = [60, 140, 250]  # degrees
distances = [8, 5, 10]     # km

# Starting point
x, y = 0, 0

x_points = [x]
y_points = [y]

for bearing, distance in zip(bearings, distances):

    # Bearing measured clockwise from North
    angle = np.radians(90 - bearing)

    dx = distance * np.cos(angle)
    dy = distance * np.sin(angle)

    x += dx
    y += dy

    x_points.append(x)
    y_points.append(y)

print(f"Final coordinates: ({x:.2f}, {y:.2f})")

straight_line_distance = np.sqrt(x**2 + y**2)
print(f"Distance from start: {straight_line_distance:.2f} km")

plt.figure(figsize=(8,6))
plt.plot(x_points, y_points, marker='o')
plt.grid(True)

for i in range(len(x_points)):
    plt.annotate(f"P{i}", (x_points[i], y_points[i]))

plt.xlabel("East-West Position (km)")
plt.ylabel("North-South Position (km)")
plt.title("GPS Navigation Simulator")
plt.axis('equal')

plt.show()
```

## Bonus: 'N' Movements

```python
n = int(input("How many movements? "))

bearings = []
distances = []

for i in range(n):
    b = float(input(f"Bearing {i+1} (degrees): "))
    d = float(input(f"Distance {i+1} (km): "))

    bearings.append(b)
    distances.append(d)

x, y = 0, 0

x_points = [0]
y_points = [0]

for bearing, distance in zip(bearings, distances):

    angle = np.radians(90 - bearing)

    dx = distance * np.cos(angle)
    dy = distance * np.sin(angle)

    x += dx
    y += dy

    x_points.append(x)
    y_points.append(y)

print(f"\nFinal Coordinates = ({x:.2f}, {y:.2f})")

plt.figure(figsize=(8,6))
plt.plot(x_points, y_points, marker='o')
plt.grid(True)

for i in range(len(x_points)):
    plt.annotate(f"P{i}", (x_points[i], y_points[i]))

plt.title("Student Navigation Route")
plt.xlabel("East-West Position")
plt.ylabel("North-South Position")
plt.axis('equal')
plt.show()
```
