```python
import matplotlib.pyplot as plt

P = 10000
r = 0.10
duration = 10

years = list(range(duration + 1))

simple = [P*(1+r*t) for t in years]
compound = [P*((1+r)**t) for t in years]

plt.plot(years, simple, marker="x", linestyle="-", label="Simple Interest")
plt.plot(years, compound, marker="o", linestyle="-", label="Compound Interest")

plt.xlabel("Years")
plt.ylabel("Amount (₦)")
plt.grid()
plt.legend()
plt.show()
```
