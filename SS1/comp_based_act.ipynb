# Simple Caesar Cipher using Modular Arithmetic

```python
# Simple Caesar Cipher using Modular Arithmetic

alphabet = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"

def encrypt(message, shift=5):
    result = ""
    for char in message.upper():
        if char in alphabet:
            x = alphabet.index(char)
            encrypted_index = (x + shift) % 26
            result += alphabet[encrypted_index]
        else:
            result += char
    return result
```

```python
def decrypt(message, shift=5):
    result = ""
    for char in message.upper():
        if char in alphabet:
            x = alphabet.index(char)
            decrypted_index = (x - shift) % 26
            result += alphabet[decrypted_index]
        else:
            result += char
    return result
```

```python
command_choice = input("Enter 'e' to encrypt or 'd' to decrypt: ")

message = input("Enter a message: ")

if command_choice.lower() == 'e':
    encrypted = encrypt(message)
    print("Encrypted message:", encrypted)
elif command_choice.lower() == 'd':
    decrypted = decrypt(message)
    print("Decrypted message:", decrypted)
else:
    print("Invalid choice. Please enter 'e' or 'd'.")
```
